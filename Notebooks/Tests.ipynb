```python
class Eleve:
    def __init__(self, nom, prenom):
        self.nom = nom
        self.prenom = prenom

    def afficher(self):
        print(f"{self.nom} {self.prenom}")

    def __repr__(self):
        return f"{self.nom} {self.prenom}"
```

```python
toto = Eleve("John", "Doe")
toto.afficher()
```

```python
toto.prenom = "Joe"
```
